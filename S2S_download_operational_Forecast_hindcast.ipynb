# Downloading the S2S_Forecast Rainfall and Tempertaure

```python
from acacia_s2s_toolkit.download_forecast import download_forecast
import os

## custom parameters
model = "ECMWF"
variable = "tp"

# 1. FIXED SYNTAX: Changed colon (:) to equals (=) for assignment
bbox_bounds = [15, 33, 3, 48] 

# 2. FIXED PATH: Removed the extra stray double quote (") inside the f-string
data_save_dir = f'/home/mesay/Desktop/ACACIA/S2S_Data/{model}/{variable}'

if variable == '2t':
    leadtime_hour = [168, 336, 504, 672] # for rainfall, this is accumulation at the end of each week
elif variable == '2t':
    ## need daily data out to 4 weeks
    leadtime_hour = list(range(0, 672+24, 24))
    print(leadtime_hour)

os.makedirs(data_save_dir, exist_ok=True)

# 3. FIXED ARGUMENTS: Removed 'location_name' entirely since it throws errors,
# and updated 'bbox_bounds' to use your custom Ethiopia boundaries [15, 33, 3, 48]
download_forecast(
    model, 
    variable, 
    fcdate=None, 
    plevs=None, 
    bbox_bounds=bbox_bounds, 
    filename=None, 
    data_save_dir=data_save_dir, 
    data_format="netcdf", 
    grid="1.5/1.5", 
    leadtime_hour=leadtime_hour,
    fc_enslags=None, 
    overwrite=False, 
    verbose=True
)
```

# Download the hindcast

```python
from acacia_s2s_toolkit.download_hindcast import download_hindcast
import numpy as np
import os
```

```python
## custom parameters
model = "ECMWF"
variable = "tp"

fcdate="20260914"

# 1. FIXED SYNTAX: Changed colon (:) to equals (=) for assignment
bbox_bounds = [15, 33, 3, 48] 

# 2. FIXED PATH: Removed the extra stray double quote (") inside the f-string
data_save_dir = f'/home/mesay/Desktop/ACACIA/S2S_Data/{model}/{variable}'

if variable == 'tp':
#    leadtime_hour = [168, 336, 504, 672] # for rainfall, this is accumulation at the end of each week
    leadtime_hour = np.arange(24,28*24,24) # for rainfall, this is accumulation at the end of each week
elif variable == '2t':
    ## need daily data out to 4 weeks
    leadtime_hour = list(range(0, 672+24, 24))
    print(leadtime_hour)

os.makedirs(data_save_dir, exist_ok=True)


download_hindcast(
    model, 
    variable, 
    fcdate=fcdate, 
    plevs=None, 
    bbox_bounds=bbox_bounds, 
    filename=None, 
    data_save_dir=data_save_dir, 
    data_format="netcdf", 
    grid="1.5/1.5", 
    leadtime_hour=leadtime_hour,
    rf_enslags=None, 
    overwrite=True, 
    verbose=True
)

download_hindcast("ECMWF","tp", fcdate=fcdate,leadtime_hour=leadtime_hour,bbox_bounds=[15, 33, 3, 48],data_save_dir=data_save_dir)
```
